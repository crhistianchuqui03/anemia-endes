```python
# ============================================================
# CARGA CORRECTA DE LOS MÓDULOS ENDES 2022
# ============================================================

import pandas as pd
import numpy as np

# RECH0 separado por punto y coma
rech0 = pd.read_csv(
    "RECH0.csv",
    sep=";",
    encoding="latin-1",
    low_memory=False
)

# RECH6 separado por coma
rech6 = pd.read_csv(
    "RECH6.csv",
    sep=",",
    encoding="latin-1",
    low_memory=False
)

print("RECH0:", rech0.shape)
print("RECH6:", rech6.shape)
```

```python
# ============================================================
# LIMPIEZA DE NOMBRES DE COLUMNAS
# ============================================================

rech0.columns = (
    rech0.columns
    .str.replace("\ufeff", "", regex=False)
    .str.replace("ï»¿", "", regex=False)
    .str.strip()
)

rech6.columns = (
    rech6.columns
    .str.replace("\ufeff", "", regex=False)
    .str.replace("ï»¿", "", regex=False)
    .str.strip()
)

print("Primeras columnas RECH0:")
print(rech0.columns[:10].tolist())

print("\nPrimeras columnas RECH6:")
print(rech6.columns[:10].tolist())
```

```python
print("ID1 en RECH0:", "ID1" in rech0.columns)
print("ID1 en RECH6:", "ID1" in rech6.columns)
```

```python
# ============================================================
# VERIFICACIÓN DEL AÑO
# ============================================================

print("Años encontrados en RECH0:")
print(rech0["ID1"].value_counts(dropna=False))

print("\nAños encontrados en RECH6:")
print(rech6["ID1"].value_counts(dropna=False))
```

```python
# ============================================================
# 4. VERIFICACIÓN DE LA LLAVE HHID
# ============================================================

print("¿HHID existe en RECH0?", "HHID" in rech0.columns)
print("¿HHID existe en RECH6?", "HHID" in rech6.columns)

print("\nCantidad de registros:")
print("RECH0:", len(rech0))
print("RECH6:", len(rech6))

print("\nCantidad de HHID únicos:")
print("RECH0:", rech0["HHID"].nunique())
print("RECH6:", rech6["HHID"].nunique())

print("\nDuplicados de HHID en RECH0:")
print(rech0["HHID"].duplicated().sum())
```

```python
# ============================================================
# 5. INTEGRACIÓN DE RECH0 + RECH6
# ============================================================

endes_2022 = pd.merge(
    rech6,
    rech0,
    on="HHID",
    how="left",
    validate="many_to_one",
    suffixes=("_RECH6", "_RECH0")
)

print("Integración realizada correctamente.")

print("\nDimensión RECH6 antes de integrar:")
print(rech6.shape)

print("\nDimensión final después de integrar:")
print(endes_2022.shape)
```

```python
# ============================================================
# 5.1 CONTROL DE LA INTEGRACIÓN
# ============================================================

sin_coincidencia = endes_2022["HV024"].isna().sum()

print("Registros de RECH6 sin coincidencia en RECH0:")
print(sin_coincidencia)
```

```python
# ============================================================
# 5.2 VERIFICACIÓN DE REGISTROS
# ============================================================

print("Registros antes de integrar:", len(rech6))
print("Registros después de integrar:", len(endes_2022))

if len(rech6) == len(endes_2022):
    print(" No se perdieron registros durante la integración.")
else:
    print(" Revisar la integración.")
```

Se integraron los módulos RECH0 y RECH6 mediante la variable HHID, conservando los registros infantiles de RECH6 e incorporando la información correspondiente del hogar. Se verificó posteriormente la correspondencia de los registros y la conservación del número total de observaciones.

```python
# ============================================================
# 6.1 VERIFICACIÓN DE DEPARTAMENTO
# ============================================================

print("Códigos encontrados en HV024:")

print(
    endes_2022["HV024"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
# ============================================================
# 6.2 REGISTROS DE PUNO
# ============================================================

registros_puno = (endes_2022["HV024"] == 21).sum()

print("Cantidad de registros de Puno:")
print(registros_puno)
```

```python
# ============================================================
# 6.4 VALIDACIÓN DEL FILTRO
# ============================================================

print("Valores de HV024 después del filtro:")
print(puno_2022["HV024"].unique())

if puno_2022["HV024"].nunique() == 1 and puno_2022["HV024"].iloc[0] == 21:
    print("✅ La base contiene únicamente registros de Puno.")
else:
    print("⚠️ Revisar el filtro de departamento.")
```

Se realizó la selección geográfica de los registros correspondientes a la región Puno mediante la variable HV024. Posteriormente, se verificó que la base resultante contenga únicamente observaciones asociadas al departamento seleccionado.

PASO 9: Depuración y construcción de la variable objetivo
Descripción
En este paso vamos a conservar únicamente los registros con una clasificación válida de anemia y crear una variable binaria que luego usará el modelo.
Según el diccionario ENDES, HC57 corresponde al nivel de anemia y sus categorías son: 1 grave, 2 moderada, 3 leve y 4 sin anemia.     Diccionario - RECH6
Para el modelo definiremos:
1 = presenta anemia
0 = no presenta anemia

```python
# ============================================================
# 9.1 DEPURACIÓN DE LA VARIABLE HC57
# ============================================================

datos_2022 = puno_6_35_2022[
    puno_6_35_2022["HC57"].isin([1, 2, 3, 4])
].copy()

print("Registros antes de depurar:", len(puno_6_35_2022))
print("Registros válidos después de depurar:", len(datos_2022))
```

```python
# ============================================================
# 9.2 CREACIÓN DE LA VARIABLE OBJETIVO
# ============================================================

datos_2022["target_anemia"] = np.where(
    datos_2022["HC57"].isin([1, 2, 3]),
    1,
    0
)

print("Variable target_anemia creada correctamente.")
```

```python
# ============================================================
# 9.3 DISTRIBUCIÓN DE LA VARIABLE OBJETIVO
# ============================================================

print("Distribución de anemia:")

print(
    datos_2022["target_anemia"]
    .value_counts()
    .rename(index={
        1: "Con anemia",
        0: "Sin anemia"
    })
)
```

```python
# ============================================================
# 9.4 DISTRIBUCIÓN PORCENTUAL
# ============================================================

distribucion = (
    datos_2022["target_anemia"]
    .value_counts()
    .rename(index={
        1: "Con anemia",
        0: "Sin anemia"
    })
    .to_frame("Frecuencia")
)

distribucion["Porcentaje"] = (
    distribucion["Frecuencia"]
    / len(datos_2022)
    * 100
).round(2)

distribucion
```

Se depuró la variable HC57, conservando únicamente los registros con clasificación válida de anemia. Posteriormente, se construyó la variable objetivo binaria target_anemia, asignando el valor 1 a los casos de anemia grave, moderada o leve, y el valor 0 a los niños sin anemia.
